# Gikomba Market Stall Performance Analysis

**Data Analytics Assignment — Week 1**

This project analyses Wanjiku's fresh-produce stall sales and stock records over 14 days using core Python. The analysis examines sales performance, weekly growth, daily sales categories, weekday and weekend performance, and stock levels to support business decisions.

```python
daily_sales = [2300, 1950, 2800, 2100, 3600, 5200, 4100,
               2500, 2200, 3100, 2400, 3900, 5600, 4700]  # 14 days, KES

days = ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"] * 2

# qty = kg in stock, reorder = reorder level (kg), cost/price = KES per kg
stock = {
    "Tomatoes": {"qty": 18, "reorder": 25, "cost": 60, "price": 90},
    "Onions": {"qty": 40, "reorder": 20, "cost": 70, "price": 100},
    "Sukuma": {"qty": 9, "reorder": 15, "cost": 30, "price": 50},
    "Potatoes": {"qty": 55, "reorder": 30, "cost": 50, "price": 75},
    "Avocado": {"qty": 12, "reorder": 10, "cost": 80, "price": 140},
}
```

## Task 1: Overall Sales Performance

I used Python's built-in functions to calculate total, average, highest and lowest daily sales. I used the list index to identify the day and week associated with the highest and lowest sales.

```python
total_sales = sum(daily_sales)
average_sales = total_sales / len(daily_sales)
highest_sales = max(daily_sales)
lowest_sales = min(daily_sales)

best_index = daily_sales.index(highest_sales)
worst_index = daily_sales.index(lowest_sales)

best_day = days[best_index]
worst_day = days[worst_index]

best_week = best_index // 7 + 1
worst_week = worst_index // 7 + 1

print(f"Total sales: KES {total_sales:,}")
print(f"Average daily sales: KES {average_sales:,.2f}")
print(f"Highest sales: KES {highest_sales:,}")
print(f"Best day: {best_day}, Week {best_week}")
print(f"Lowest sales: KES {lowest_sales:,}")
print(f"Worst day: {worst_day}, Week {worst_week}")
```

## Task 2: Weekly Sales Totals and Growth

I created a reusable function called week_totals() that uses list slicing to separate the first seven days from the next seven days. I then calculated the percentage change from Week 1 to Week 2.

```python
def week_totals(sales):
    """Return the total sales for Week 1 and Week 2."""
    week1 = sum(sales[:7])
    week2 = sum(sales[7:])
    return week1, week2


week1_total, week2_total = week_totals(daily_sales)

growth_percentage = ((week2_total - week1_total) / week1_total) * 100

print(f"Week 1 total: KES {week1_total:,}")
print(f"Week 2 total: KES {week2_total:,}")
print(f"Week-on-week growth: {growth_percentage:.2f}%")
```

## Task 3: Daily Sales Classification

I created classify_day() to classify sales as Strong, Normal or Weak based on the overall daily average. I used a dictionary to count the number of days in each category.

```python
def classify_day(amount, average):
    """Classify daily sales based on 20% above or below the average."""
    if amount > average * 1.20:
        return "Strong"
    elif amount < average * 0.80:
        return "Weak"
    else:
        return "Normal"


category_counts = {"Strong": 0, "Normal": 0, "Weak": 0}

for amount in daily_sales:
    category = classify_day(amount, average_sales)
    category_counts[category] += 1

print("Daily sales classification:")
for category, count in category_counts.items():
    print(f"{category}: {count} days")
```

## Task 4: Weekday and Weekend Performance

I separated weekday sales (Monday to Friday) from weekend sales (Saturday and Sunday). I calculated the average daily sales for each group and the percentage by which weekend average sales exceed weekday average sales.

```python
weekday_sales = []
weekend_sales = []

for i in range(len(daily_sales)):
    if days[i] in ["Mon", "Tue", "Wed", "Thu", "Fri"]:
        weekday_sales.append(daily_sales[i])
    else:
        weekend_sales.append(daily_sales[i])

weekday_average = sum(weekday_sales) / len(weekday_sales)
weekend_average = sum(weekend_sales) / len(weekend_sales)

weekend_difference = (
    (weekend_average - weekday_average) / weekday_average
) * 100

print(f"Weekday average: KES {weekday_average:,.2f}")
print(f"Weekend average: KES {weekend_average:,.2f}")
print(f"Weekend advantage: {weekend_difference:.2f}%")
```

## Task 5: Stock Reorder and Gross Margin Analysis

I used a function to identify products whose stock quantity is at or below their reorder level. I also used the dictionary items() method to calculate each product's gross margin percentage using its selling price and cost price.

```python
def products_to_reorder(inventory):
    """Return product names with stock at or below reorder level."""
    reorder_list = []

    for product, details in inventory.items():
        if details["qty"] <= details["reorder"]:
            reorder_list.append(product)

    return reorder_list


reorder_list = products_to_reorder(stock)

print("Products requiring reorder:")
for product in reorder_list:
    print(product)

print("\nGross margin percentages:")

gross_margins = {}

for product, details in stock.items():
    margin = (
        (details["price"] - details["cost"]) / details["price"]
    ) * 100

    gross_margins[product] = margin
    print(f"{product}: {margin:.2f}%")
```

## Task 6: Stall Performance Report

I combined the calculated results into a formatted report using f-strings. I used thousands separators for KES amounts and two decimal places for percentages and averages to make the report easy to read.

```python
print("=" * 45)
print("       STALL PERFORMANCE REPORT")
print("          GIKOMBA MARKET")
print("=" * 45)

print(f"Total sales (14 days):    KES {total_sales:,}")
print(f"Average daily sales:      KES {average_sales:,.2f}")

print("\nSALES PERFORMANCE")
print(f"Highest sales:            KES {highest_sales:,}")
print(f"Best day:                 {best_day}, Week {best_week}")
print(f"Lowest sales:             KES {lowest_sales:,}")
print(f"Worst day:                {worst_day}, Week {worst_week}")

print("\nWEEKLY COMPARISON")
print(f"Week 1 sales:              KES {week1_total:,}")
print(f"Week 2 sales:              KES {week2_total:,}")
print(f"Week-on-week growth:      {growth_percentage:.2f}%")

print("\nDAY CLASSIFICATION")
for category, count in category_counts.items():
    print(f"{category:<12}: {count} days")

print("\nWEEKDAY VS WEEKEND")
print(f"Weekday average:           KES {weekday_average:,.2f}")
print(f"Weekend average:           KES {weekend_average:,.2f}")
print(f"Weekend advantage:        {weekend_difference:.2f}%")

print("\nSTOCK MANAGEMENT")
print(f"Products to reorder: {', '.join(reorder_list)}")

print("\nGROSS MARGINS")
for product, margin in gross_margins.items():
    print(f"{product:<12}: {margin:.2f}%")

print("=" * 45)
print("             END OF REPORT")
print("=" * 45)
```
